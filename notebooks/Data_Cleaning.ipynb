```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
import joblib
```

```python
df = pd.read_csv("ai4i2020.csv")
df
```

```python
df.shape
```

```python
df.describe()
```

```python
df.info()
```

```python
df['Machine failure'].unique()
```

```python
df.drop(columns=['UDI', 'Product ID'], inplace=True)
df.head()
```

```python
df['Type'].unique()
```

```python
df = pd.get_dummies(df, columns=['Type'], drop_first=True)
df.head()
```

```python
df['Air temperature [K]'] = np.clip(df['Air temperature [K]'], 250, 400)
df['Process temperature [K]'] = np.clip(df['Process temperature [K]'], 250, 400)
df['Rotational speed [rpm]'] = np.clip(df['Rotational speed [rpm]'], 0, 3000)
df['Torque [Nm]'] = np.clip(df['Torque [Nm]'], 0, 100)
df['Tool wear [min]'] = np.clip(df['Tool wear [min]'], 0, 300)
```

```python
failure_cols = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']
df.drop(columns=failure_cols, inplace=True)
```

```python
df.isnull().sum()
```

```python
df.fillna(df.median(), inplace=True)
```

```python
num_cols = [
    'Air temperature [K]',
    'Process temperature [K]',
    'Rotational speed [rpm]',
    'Torque [Nm]',
    'Tool wear [min]'
]
```

```python
scaler = StandardScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
joblib.dump(scaler, 'scaler.pkl')
```
